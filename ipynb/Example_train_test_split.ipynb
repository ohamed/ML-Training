## Example of using train_test_split from scikit-learn

### Import libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from pandas.core.common import random_state
```

### Load the dataset

```python
df_sal = pd.read_csv("data/Salary_Data.csv")
```

```python
df_sal.shape
```

### Features / target variable

```python
X =df_sal['YearsExperience'] ## X is the independent variable (Feature)
y =df_sal['Salary'] ## y is the dependent variable
```

### Splitting dataset into test/train
- Note that setting the "random_state = 0" in train_test_split ensures that the data splitting is reproducible each time you run the code.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```
